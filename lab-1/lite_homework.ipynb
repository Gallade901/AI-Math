Сырой текстовый корпус WikiText-2 нормализуется, очищается от пробелов и сегментируется на предложения с помощью функции `preprocess_text`.

```python
def preprocess_text(
    texts: List[str],
    lowercase: bool = True,
    remove_whitespace: bool = True,
    split_sentences: bool = False,
) -> List[str]:
    result = []
    # """
    # Предобрабатывает корпус текстов.

    # texts : List[str]
    #     Список исходных текстов.
    # lowercase : bool
    #     Приводить ли текст к нижнему регистру.
    # remove_whitespace : bool
    #     Удалять ли лишние пробелы.
    # split_sentences : bool
    #     Разбивать ли тексты на отдельные предложения.

    # Returns
    # -------
    # List[str]
    #     Список предобработанных строк.
    # """
    for text in texts:
        if lowercase:
            text = text.lower()

        if remove_whitespace:
            text = re.sub(r"\s+", " ", text).strip()

        if split_sentences:
            parts = re.split(r"(?<=[.!?])\s+", text)
        else:
            parts = [text]
            
        for part in parts:
            part = part.strip()
            if len(part) > 5:
                result.append(part)

    return result
    # ВАШ КОД ЗДЕСЬ
    # raise NotImplementedError("Реализуйте предобработку текста")
```

BPE строит словарь, итеративно объединяя частые пары байтов. Настроим параметры токенизатора и обучим токенизатор.


Прошу дипсик пояснить подробнее в данном контексте что такое BPE токенизатор, нормализатор NFC, пре-токенизатор и что такоt BpeTrainer - https://chat.deepseek.com/share/f4uzilet7uzv5dfves

```python
def train_bpe_tokenizer(
    texts: List[str],
    vocab_size: int,
    special_tokens: List[str] = None,
) -> Tokenizer:
    """
    Обучает BPE-токенизатор на корпусе текстов.

    texts : List[str]
        Корпус текстов для обучения токенизатора.
    vocab_size : int
        Размер результирующего словаря токенизатора.
    special_tokens : List[str], optional
        Список специальных токенов (по умолчанию [PAD], [UNK], [BOS], [EOS]).

    Returns
    -------
    Tokenizer
        Обученный объект BPE-токенизатора.
    """
    if special_tokens is None:
        special_tokens = ["[PAD]", "[UNK]", "[BOS]", "[EOS]"]

    # ВАШ КОД ЗДЕСЬ
    # 1. Инициализируйте пустой Tokenizer с BPE моделью
    # 2. Настройте нормализатор (NFC, Lowercase)
    # 3. Настройте пре-токенизатор ByteLevel
    # 4. Настройте декодер ByteLevel
    # 5. Обучите токенизатор с помощью BpeTrainer
    tokenizer = Tokenizer(models.BPE(unk_token="[UNK]"))

    tokenizer.normalizer = normalizers.Sequence([
        normalizers.NFC(),#Нормализация разных символов к одному
        normalizers.Lowercase(),
    ])

    tokenizer.pre_tokenizer = pre_tokenizers.ByteLevel(add_prefix_space=False) #Кодируем строки в байты и пробелы в спец символы
    tokenizer.decoder = decoders.ByteLevel()

    trainer = trainers.BpeTrainer( #Настройка
        vocab_size=vocab_size,
        special_tokens=special_tokens,
        min_frequency=2, #Минимальное количество встреч для слияния
        show_progress=False,
    )

    tokenizer.train_from_iterator(texts, trainer=trainer) #Запуск

    tokenizer.post_processor = processors.TemplateProcessing(#Добавление служебных токенов
        single="[BOS] $A [EOS]", #входные токены после BPE:  ["При", "вет", "Ġмир"]результат после шаблона:   ["[BOS]", "При", "вет", "Ġмир", "[EOS]"] ids:[id_BOS, ..., id_EOS]
        special_tokens=[
            ("[BOS]", tokenizer.token_to_id("[BOS]")),
            ("[EOS]", tokenizer.token_to_id("[EOS]")),
        ],
    )

    return tokenizer
```

```python
def show_tokenization(
    tokenizer: Tokenizer,
    name: str,
    examples: List[str],
) -> None:
    """
    Выводит результаты токенизации примеров.

    tokenizer : Tokenizer
        Обученный токенизатор.
    name : str
        Название токенизатора для вывода.
    examples : List[str]
        Список тестовых примеров.
    """
    # ВАШ КОД ЗДЕСЬ
    # Для каждого примера выведите строковое представление токенов и их ID
    print(f"\n=== {name} ===")
    for example in examples:
        encoded = tokenizer.encode(example)
        print(f"\nТекст: {example}")
        print(f"Токены: {encoded.tokens}")
        print(f"ID: {encoded.ids}")
```

>
> Как размер словаря токенизатора влияет на степень фрагментации слов при токенизации? Каковы преимущества и недостатки использования большого размера словаря в реальных языковых моделях?
>
> ***Ответ:*** При увеличении словаря у BPE появляется больше готовых подслов и целых слов, поэтому среднее число токенов на слово обычно уменьшается, а фрагментация снижается. Это уменьшает длину последовательностей и, следовательно, число шагов внимания для одного и того же текста. Обратная сторона - больший размер embedding-матрицы и выходной матрицы модели, рост памяти и числа параметров, а также потенциально более разреженное использование редких токенов. Поэтому размер словаря является компромиссом между компактностью последовательности и стоимостью модели.

```python
def compare_tokenizers(
    tokenizer_a,
    tokenizer_b,
    name_a: str,
    name_b: str,
    texts: List[str],
    category: str,
) -> Dict[str, List[float]]:
    """
    Сравнивает два токенизатора по скорости сжатия текста.
    """
    # ВАШ КОД ЗДЕСЬ: Вычислите количество токенов на 100 символов и коэффициент сжатия для обоих токенизаторов
    def encode_len(tokenizer, text):
        try:
            encoded = tokenizer.encode(text, add_special_tokens=False)
        except TypeError:
            encoded = tokenizer.encode(text).ids

        if hasattr(encoded, "ids"):
            encoded = encoded.ids

        return len(encoded)

    tokens_per_100_a = []
    tokens_per_100_b = []
    compression_a = []
    compression_b = []

    for text in texts:
        char_len = max(len(text), 1)

        n_a = encode_len(tokenizer_a, text)
        n_b = encode_len(tokenizer_b, text)

        tokens_per_100_a.append(n_a / char_len * 100)
        tokens_per_100_b.append(n_b / char_len * 100)

        compression_a.append(char_len / max(n_a, 1))
        compression_b.append(char_len / max(n_b, 1))

    return {
        "name_a": name_a,
        "name_b": name_b,
        "category": category,
        "tokens_per_100_a": float(np.mean(tokens_per_100_a)),
        "tokens_per_100_b": float(np.mean(tokens_per_100_b)),
        "compression_a": float(np.mean(compression_a)),
        "compression_b": float(np.mean(compression_b)),
    }
```

Для количественной оценки качества языковой модели после подмены токенизатора рассчитывается перплексия (perplexity) с помощью скользящего окна контекста.

```python
def compute_perplexity(
    model,
    tokenizer,
    texts: List[str],
    max_length: int = 256,
    stride: int = 128,
    device: str = "cuda" if torch.cuda.is_available() else "cpu"
) -> float:
    """
    Вычисляет перплексию предобученной модели на тестовом корпусе текстов.
    """
    total_loss = 0.0
    total_len = 0
    device_str = str(device)
    
    for text in texts:
        if isinstance(tokenizer, Tokenizer):
            enc = tokenizer.encode(text)
            input_ids = torch.tensor([enc.ids]).to(device_str)
        else:
            enc = tokenizer(text, return_tensors="pt")
            input_ids = enc.input_ids.to(device_str)
            
        seq_len = input_ids.size(1)
        if seq_len <= 1:
            continue
            
        prev_end_loc = 0
        for begin_loc in range(0, seq_len, stride):
            end_loc = min(begin_loc + max_length, seq_len)
            trg_len = end_loc - prev_end_loc
            
            inputs = input_ids[:, begin_loc:end_loc]
            targets = inputs.clone()
            targets[:, :-trg_len] = -100
            
            # ВАШ КОД ЗДЕСЬ: Выполните прямой проход модели (forward) с labels=targets
            # Получите значение loss и прибавьте к total_loss величину (loss * trg_len)
            # Не забудьте использовать torch.no_grad()
            with torch.no_grad():
                outputs = model(inputs, labels=targets)
                loss = outputs.loss

            total_loss += loss.item() * trg_len
        
            
            total_len += trg_len
            prev_end_loc = end_loc
            if end_loc == seq_len:
                break
                
    return math.exp(total_loss / total_len) if total_len > 0 else float("inf")
```

> ***Ответ:*** Перплексия - это экспонента от средней кросс-энтропии. Она показывает, насколько модель удивлена текстом: чем меньше, тем увереннее модель предсказывает следующий токен. Для мягкого управления генерацией (например, стимулирования модели использовать определенные слова) применяется механизм Keyword Boosting.

Прошу дипсик объяснить более подробно задание в данном контексте - https://chat.deepseek.com/share/4eue2j87rhhrub8io8

```python
class KeywordGuidanceLogitsProcessor(LogitsProcessor):
    """
    LogitsProcessor, «подталкивающий» модель к использованию слов
    из заданного списка (keyword boosting).
    """

    def __init__(self, tokenizer, keywords: List[str], boost: float = 5.0):
        self.tokenizer = tokenizer
        self.boost = boost
        self.keyword_ids = set()

        for keyword in keywords:
            try:
                ids = tokenizer.encode(keyword, add_special_tokens=False)
            except TypeError:
                ids = tokenizer.encode(keyword).ids

            if hasattr(ids, "ids"):
                ids = ids.ids

            self.keyword_ids.update(ids)

    def __call__(
        self,
        input_ids: torch.LongTensor,
        scores: torch.FloatTensor,
    ) -> torch.FloatTensor:
        if self.keyword_ids:
            idx = torch.tensor(
                list(self.keyword_ids),
                device=scores.device,
                dtype=torch.long,
            )
            scores[:, idx] += self.boost

        return scores
```
